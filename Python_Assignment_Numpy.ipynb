**Array Operations**

```python
import numpy as np

# Temperatures in Celsius
temperatures_w1 = np.array([25, 28, 30, 27, 26, 32, 29])

# Convert Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Celsius:", temperatures_w1)
print("Temperatures in Fahrenheit:", temperatures_f)


# Calculate maximum, minimum, and mean
print("Maximum temperature:", np.max(temperatures_w1), "°C")
print("Minimum temperature:", np.min(temperatures_w1), "°C")
print("Mean temperature:", np.mean(temperatures_w1), "°C")



```

**Array Slicing and Indexing**

```python
import numpy as np

temperatures_w1 = np.array([25, 28, 30, 27, 29, 31, 26])

# First three days of the week (Monday to Wednesday)
first_three_days = temperatures_w1[:3]
print("First three days:", first_three_days)


# Weekend (Saturday and Sunday)
weekend = temperatures_w1[-2:]
print("Weekend temperatures:", weekend)

# Middle three days of the week (Wednesday to Friday)
middle_three_days = temperatures_w1[2:5]
print("Middle three days:", middle_three_days)



```

**Inspect and Slice the 2D Array**

```python
import numpy as np

# Create a 2D array for two weeks
temperatures = np.array([
                       [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
                       [19.2, 22.5, 21.3, 24.0, 23.5, 26.2, 20.7]
])

# 1. Inspect the array
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)

# 2. Extract temperatures for each week
week1 = temperatures[0, :]
week2 = temperatures[1, :]


print("\nWeek 1 temperatures:", week1)
print("Week 2 temperatures:", week2)

# 3. Extract weekend temperatures (last two days)
weekends = temperatures[:, -2:]

print("\nWeekend temperatures for both weeks:")
print(weekends)


# 4. Extract weekends separately
week1_weekend = temperatures[0, -2:]
week2_weekend = temperatures[1, -2:]


print("\nWeek 1 weekend:", week1_weekend)
print("Week 2 weekend:", week2_weekend)





```

**Pandas Series**

** Creating Pandas Series**

```python
import pandas as pd


marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

**Indexing and Slicing**

```python
import pandas as pd

# Create a Pandas Series
marks = pd.Series(
        [95, 92, 89, 85, 80],
        index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

# 1. Access the mark of the 1st rank student using integer index position
print("Mark of 1st rank student:")
print(marks[0])

# 2. Retrieve marks of the top 3 ranks using loc
print("\nMarks of top 3 ranks using loc:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])


# 3. Retrieve the mark of the 3rd-ranked student using iloc
print("\nMark of 3rd-ranked student using iloc:")
print(marks.iloc[2])


# 4. Filter ranks where marks are greater than 90 using a Boolean mask
print("\nRanks with marks greater than 90:")
print(marks[marks > 90])

```

**Manipulating Series**

```python
import pandas as pd

# Create the Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

# 1. Modify the mark of the 1st rank student to 100
marks['Rank1'] = 100
print("Updated Marks:")
print(marks)


# 2. Remove the entry corresponding to the last rank
marks = marks.drop('Rank5')
print("\nMarks after removing the last rank:")
print(marks)


# 3. Compute CGPA by dividing each mark by 10
cgpa = marks / 10
print("\nCGPA:")
print(cgpa)

```

**Pandas DataFrame**

**Creating Pandas DataFrame**

```python
import pandas as pd

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
     'ProductCategory': [
         'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
         'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'

     ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print(transactions)
```

**Data Exploration**

```python
import pandas as pd

# 1. Display the DataFrame
print("Transactions DataFrame:")
print(transactions)

# Display the first 5 rows
print("\nHead of the DataFrame:")
print(transactions.head())

# Display the last 5 rows
print("\nTail of the DataFrame:")
print(transactions.tail())

# Display basic information
print("\nBasic Information:")
transactions.info()

# Display shape
print("\nShape of DataFrame:")
print(transactions.shape)

# Display column names
print("\nColumn Names:")
print(transactions.columns)


# Display data types
print("\nData Types:")
print(transactions.dtypes)

# 2. Display ProductCategory and Amount columns
print("\nProductCategory and Amount:")
print(transactions[['ProductCategory', 'Amount']])

# 3. Retrieve the last 3 columns using iloc
print("\nLast 3 Columns:")
print(transactions.iloc[:, -3:])


# 4. Filter rows where Region is North and Amount > 200
print("\nNorth Region with Amount Greater than 200:")
print(transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
])


# 5. Value counts for ProductCategory
print("\nProduct Category Value Counts:")
print(transactions['ProductCategory'].value_counts())

# 6. Unique values in Region
print("\nUnique Regions:")
print(transactions['Region'].unique())


# 7. Mean Amount for each Region
print("\nMean Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())

```

**Manipulating the DataFrame**

```python
import pandas as pd

# Modify Amount for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print("After modifying Amount:")
print(transactions)


# Add a new Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

print("\nAfter adding Discount column:")
print(transactions)

# Remove the row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

print("\nAfter removing TransactionID 109:")
print(transactions)

# Delete the Discount column
transactions.drop(columns=['Discount'], inplace=True)

print("\nAfter deleting Discount column:")
print(transactions)


```
